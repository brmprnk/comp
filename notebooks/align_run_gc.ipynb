```python
txt_run = "input/run.txt"
txt_gc = "input/run_gc.txt"
```

```python
# Make sure that the lines in each file contain the same ID, otherwise remove the lines
# that do not match.
run = []
gc = []
with open(txt_run) as f:
    for line in f:
        run.append(line)
with open(txt_gc) as f:
    for line in f:
        gc.append(line)

len(run), len(gc)
```

```python
run_aligned = []
for i in range(len(gc)):
    gc_id = gc[i].split("/")[-1].split(".")[0]

    # Find the string in run that contains the same ID
    for j in range(len(run)):
        run_id = run[j].split("/")[-1].split(".")[0]
        if gc_id == run_id:
            # remove the newline character from gc
            removed_newline = run[j].rstrip("\n")

            run_aligned.append(removed_newline)
            break

len(run_aligned), len(gc)
```

```python
# Write the aligned run to a file
with open("input/run.txt", "w") as f:
    for line in run_aligned:
        f.write(line + "\n")
```
