```python
import pandas as pd
import numpy as np
```

```python
biomart = pd.read_csv('beds/biomart_10kb.bed', sep='\t', header=None, names=['chrom', 'start', 'end', 'name', 'overlap', 'overlap_encode'])
to_be_removed = np.load('accessory_files/blacklist_and_std_outlier_and_chrX_indices.npy')
biomart_hg19 = pd.read_csv('beds/biomart_10kb_hg19.bed', sep='\t', header=None, names=['chrom', 'start', 'end', 'name', 'strand', 'gc'])
biomart
```

```python
biomart_filtered = biomart_hg19[~biomart_hg19['name'].isin(biomart.iloc[to_be_removed]['name'].values)]
biomart_filtered['chrom'].value_counts()
```

```python
names_to_be_removed = biomart.iloc[to_be_removed]['name'].values
print("Number of names to be removed:", len(names_to_be_removed), "from total of", biomart.shape[0], "selected by indices in to_be_removed.npy:", to_be_removed.shape)
biomart_hg19_filtered = biomart_hg19[~biomart_hg19['name'].isin(names_to_be_removed)]
print(biomart_hg19.shape[0] -  biomart_hg19_filtered.shape[0])

# Find the indices of the removed regions in the hg19 biomart
removed_indices_hg19 = biomart_hg19.index[biomart_hg19['name'].isin(names_to_be_removed)].tolist()
print(removed_indices_hg19)
print(len(removed_indices_hg19))
np.save('accessory_files/blacklist_and_std_outlier_and_chrX_indices_hg19.npy', np.array(removed_indices_hg19))
```
