```python
from pathlib import Path

import pandas as pd
```

```python
csv_dir = Path("../extracted_features/hg38_1mb_filtered_bins")
files = sorted(csv_dir.glob("*.csv"))

bin_ids = [f.name.split(".")[0] for f in files]
print("The number of samples processed for bin features is:", len(bin_ids))
```

```python
summaries = []

for f in files:
    try:
        df_head = pd.read_csv(f, nrows=0)
        summaries.append(
            {
                "file": f.name,
                "n_cols": len(df_head.columns),
                "columns": tuple(df_head.columns),
            }
        )
    except Exception as e:
        summaries.append({"file": f.name, "error": str(e)})

summary_df = pd.DataFrame(summaries)
summary_df.head()
```

```python
# Check that all files have the same columns
print(summary_df["n_cols"].value_counts())
print(summary_df["columns"].value_counts())
```

```python
def count_rows(file, chunksize=100_000):
    total = 0
    for chunk in pd.read_csv(file, chunksize=chunksize):
        total += len(chunk)
    return total


row_counts = []

for f in files:
    try:
        row_counts.append({"file": f.name, "n_rows": count_rows(f)})
    except Exception as e:
        row_counts.append({"file": f.name, "error": str(e)})

rows_df = pd.DataFrame(row_counts)
rows_df.head()
```

```python
rows_df["n_rows"].describe()
rows_df["n_rows"].value_counts()
```

```python
# Compare bin_samples to cris_bam.txt - to check all samples were processed

with open("../accessory_files/cris_bam.txt") as f:
    cris_bam_ids = [line.replace("cris/", "").split(".")[0] for line in f.readlines()]

# print(cris_bam_ids)
missing = list(set(cris_bam_ids) - set(bin_ids))
print("Number of missing samples:", len(missing))
print("Missing samples are:", missing)
```
